```python
# Import necessary libraries

import pandas as pd
import numpy as np

# For calculating similarity
from sklearn.metrics.pairwise import cosine_similarity

# For splitting dataset and evaluation
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
from math import sqrt
```

```python
# Load movies dataset
movies = pd.read_csv('/content/drive/MyDrive/CSE-Data-Mining-Warehouse-Lab-426/Recommendation System Project/movies.csv')

# Load ratings dataset
ratings = pd.read_csv('/content/drive/MyDrive/CSE-Data-Mining-Warehouse-Lab-426/Recommendation System Project/ratings.csv')
```

```python
# Show first 5 rows of movies dataset
movies.head()
```

```python
# Show first 5 rows of ratings dataset
ratings.head()
```

```python
# Display dataset shape
print("Movies Dataset Shape:", movies.shape)
print("Ratings Dataset Shape:", ratings.shape)
```

```python
# Create user-item interaction matrix

user_item_matrix = ratings.pivot_table(
    index='userId',
    columns='movieId',
    values='rating'
)

# Replace missing values with 0
user_item_matrix = user_item_matrix.fillna(0)

# Display matrix
user_item_matrix.head()
```

```python
# Calculate cosine similarity between movies

item_similarity = cosine_similarity(user_item_matrix.T)

# Convert similarity matrix into dataframe
item_similarity_df = pd.DataFrame(
    item_similarity,
    index=user_item_matrix.columns,
    columns=user_item_matrix.columns
)

# Display similarity matrix
item_similarity_df.head()
```

```python
# Function to recommend movies

def recommend_movies(movie_id, top_n=5):

    # Get similarity scores
    similar_scores = item_similarity_df[movie_id]

    # Sort similarity scores
    similar_movies = similar_scores.sort_values(ascending=False)

    # Remove the same movie
    similar_movies = similar_movies.iloc[1:top_n+1]

    return similar_movies
```

```python
# Example movie ID
movie_id = 1

# Get recommendations
recommended_movies = recommend_movies(movie_id)

# Display recommended movie IDs
recommended_movies
```

```python
# Display recommended movie titles

for movie_id in recommended_movies.index:

    movie_title = movies[movies['movieId'] == movie_id]['title'].values[0]

    print(movie_title)
```

```python
# Get unique users
users = ratings['userId'].unique()[:5]

# Generate recommendations for first 5 users

for user in users:

    print("\n==============================")
    print(f"Recommendations for User {user}")
    print("==============================")

    # Get movies rated by the user
    user_movies = ratings[ratings['userId'] == user]

    # Get highest rated movie
    top_movie = user_movies.sort_values(
        by='rating',
        ascending=False
    ).iloc[0]

    movie_id = top_movie['movieId']

    # Recommend similar movies
    recommendations = recommend_movies(movie_id)

    for rec_movie_id in recommendations.index:

        movie_title = movies[
            movies['movieId'] == rec_movie_id
        ]['title'].values[0]

        print(movie_title)
```

```python
# Split dataset into train and test data

train_data, test_data = train_test_split(
    ratings,
    test_size=0.2,
    random_state=42
)

print("Training Data Shape:", train_data.shape)
print("Testing Data Shape:", test_data.shape)
```

```python
# Function to predict rating

def predict_rating(user_id, movie_id):

    try:

        # Get movies rated by user
        user_ratings = user_item_matrix.loc[user_id]

        # Get similarity scores
        similarity_scores = item_similarity_df[movie_id]

        # Calculate weighted average
        numerator = np.dot(user_ratings, similarity_scores)

        denominator = similarity_scores.sum()

        prediction = numerator / denominator

        return prediction

    except:

        return 0
```

```python
# Store actual and predicted ratings

actual_ratings = []
predicted_ratings = []

# Predict ratings for test data

for row in test_data.itertuples():

    actual = row.rating
    predicted = predict_rating(row.userId, row.movieId)

    actual_ratings.append(actual)
    predicted_ratings.append(predicted)

# Calculate RMSE

rmse = sqrt(mean_squared_error(
    actual_ratings,
    predicted_ratings
))

print("RMSE Value:", rmse)
```
